```python
from bs4 import BeautifulSoup
import re
from nltk.corpus import stopwords
def clean_text(raw_text):


    text = BeautifulSoup(raw_text, "html.parser").get_text()

    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    text = re.sub(r'[^a-zA-Z\s]', '', text)

    words = text.lower().split()

    stop_words = set(stopwords.words('english'))
    meaningful_words = [w for w in words if not w in stop_words]

    cleaned_text = " ".join(meaningful_words).strip()

    return cleaned_text
```

```python
new_df['tags'] = new_df['tags'].apply(clean_text)
```

```python
new_df.head()
```

```python
from nltk.stem import WordNetLemmatizer
wl=WordNetLemmatizer()
```

```python
def lemmatize_words(text):
    return " ".join([wl.lemmatize(word) for word in text.split()])
```

```python
new_df['tags']=new_df['tags'].apply(lambda x : lemmatize_words(x))
```

```python
new_df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test= train_test_split(new_df['tags'],new_df['rating'],test_size=.20)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000 , stop_words = 'english')
```

```python
X_train_cv= cv.fit_transform(X_train).toarray()
X_test_cv=cv.transform(X_test).toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf=TfidfVectorizer()
```

```python
X_train_tf= tf.fit_transform(X_train).toarray()
X_test_tf=tf.transform(X_test).toarray()
```

```python
X_test_cv
```

```python
X_train_tokenized = X_train.apply(lambda text: text.split())
X_test_tokenized = X_test.apply(lambda text: text.split())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

# Train Logistic Regression model with CountVectorizer data
lg_model_cv = LogisticRegression().fit(X_train_cv, y_train)

# Make predictions on the test data
y_pred_cv = lg_model_cv.predict(X_test_cv)

# Evaluate the model
print("Confusion Matrix (CountVectorizer):")
print(confusion_matrix(y_test, y_pred_cv))
print("\nClassification Report (CountVectorizer):")
print(classification_report(y_test, y_pred_cv))
print("\nAccuracy Score (CountVectorizer):", accuracy_score(y_test, y_pred_cv))
```

```python
def predict_sentiment(text):
    cleaned_text = clean_text(text)
    # Use the fitted CountVectorizer to transform the text
    text_vector = cv.transform([cleaned_text])
    # Make sure the vector is in the correct format for the model
    text_vector = text_vector.reshape(1, -1)
    prediction = lg_model_cv.predict(text_vector)
    return prediction[0]

# Example usage:
test_sentence = "I like this book."
predicted_sentiment = predict_sentiment(test_sentence)

if predicted_sentiment == 1:
    print("Predicted Sentiment: Positive")
else:
    print("Predicted Sentiment: Negative")

test_sentence_2 = "This book was terrible."
predicted_sentiment_2 = predict_sentiment(test_sentence_2)

if predicted_sentiment_2 == 1:
    print("Predicted Sentiment: Positive")
else:
    print("Predicted Sentiment: Negative")
```
